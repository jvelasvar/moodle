## 1.1 El entorno de MATLAB

### Concepto
MATLAB tiene cuatro zonas principales:

| Zona | Para qué sirve |
|---|---|
| **Command Window** | Escribir y ejecutar instrucciones una a una |
| **Workspace** | Ver las variables que existen en memoria |
| **Editor / Live Editor** | Escribir scripts (`.m`) y Live Scripts (`.mlx`) |
| **Current Folder** | La carpeta donde MATLAB busca y guarda archivos |

**Comandos esenciales**

| Comando | Acción |
|---|---|
| `help nombre` / `doc nombre` | Ayuda rápida / documentación completa |
| `clc` | Limpia la pantalla (no borra variables) |
| `clear x` / `clear` | Borra la variable `x` / todas las variables |
| `who`, `whos` | Lista las variables / con tamaño y tipo |
| `close all` | Cierra todas las figuras |
| `;` | Al final de una línea, oculta el resultado |
| `%` | Inicia un comentario |
| `ans` | Guarda el último resultado sin nombre |

> 💡 **Analogía:** el *Workspace* es tu **mesa de laboratorio**: todo lo que calculas queda sobre ella hasta que lo retiras con `clear`.

### Ejemplos

```matlab
% Ejemplo 1.1.1: MATLAB como calculadora
2 + 3*4          % sin punto y coma, el resultado se guarda y se muestra en "ans"
ans / 2          % "ans" se puede reutilizar en el siguiente cálculo
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `2 + 3*4`: MATLAB respeta la jerarquía de operaciones (primero `*`, luego `+`), por lo que da 14. Como no termina en `;`, imprime el resultado.
  - Al no asignarlo a ninguna variable, MATLAB lo guarda automáticamente en la variable especial `ans` (*answer*).
- `ans / 2`: usa el valor guardado en `ans` (14) y lo divide entre 2. El nuevo resultado (7) reemplaza a `ans`.
- `%`: todo lo que está a la derecha es un **comentario**; MATLAB lo ignora.

</details>

```matlab
% Ejemplo 1.1.2: variables con nombre y el uso del punto y coma
R = 220;          % resistencia [ohm]; el ";" oculta la salida
Ic = 0.015;       % corriente [A]
V = R*Ic          % tensión por la ley de Ohm [V]; sin ";" se muestra
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `R = 220;`: crea la variable `R` con el valor 220. El `=` significa **asignación** ("guarda en"), no igualdad matemática. El `;` evita que se imprima.
- `Ic = 0.015;`: crea la variable `Ic` (corriente en amperios). MATLAB distingue mayúsculas y minúsculas: `Ic` e `ic` son variables distintas.
- `V = R*Ic`: multiplica las dos variables y guarda el resultado en `V`. Sin `;`, se muestra `V = 3.3000`.

</details>

```matlab
% Ejemplo 1.1.3: inspeccionar y limpiar el workspace
who               % nombres de las variables existentes
whos R Ic V       % tamaño, memoria y tipo de R, Ic y V
clear Ic          % borra únicamente la variable Ic
existe_Ic = exist('Ic')   % 0 significa que Ic ya no existe
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `who`: lista los nombres de las variables que existen en el workspace.
- `whos R Ic V`: muestra información detallada de esas tres variables: `Size` (tamaño, por ejemplo `1x1`), `Bytes` (memoria) y `Class` (tipo, por ejemplo `double`). Sin argumentos, `whos` muestra todas.
- `clear Ic`: borra **solo** la variable `Ic`; las demás se conservan.
- `existe_Ic = exist('Ic')`: `exist` recibe el nombre de una variable **como texto** (entre comillas) y devuelve 0 si no existe. Da 0 porque la acabamos de borrar.

</details>

```matlab
% Ejemplo 1.1.4: pedir ayuda sobre una función
help sqrt         % muestra la ayuda rápida de sqrt
% doc sqrt        % (en MATLAB) abre la documentación completa con ejemplos
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `help sqrt`: muestra en la Command Window una ayuda resumida de la función `sqrt` (raíz cuadrada): qué hace, cómo se llama y funciones relacionadas.
- `% doc sqrt`: está comentada. Si quitas el `%`, `doc` abre la documentación completa con ejemplos y gráficas. Úsala siempre que no recuerdes los parámetros de una función.

</details>

## 1.2 Variables y tipos de datos

### Concepto
En MATLAB **todo es una matriz**: un número es una matriz $1\times1$. Los tipos más usados en este curso son:

| Tipo | Ejemplo | Uso en Señales y Sistemas |
|---|---|---|
| `double` | `x = 3.7` | Muestras de señales, parámetros |
| complejo | `z = 3 + 4j` | Fasores, exponenciales complejas, Fourier, polos |
| `char` / `string` | `'texto'` / `"texto"` | Títulos y etiquetas |
| `logical` | `t >= 0` | Máscaras: escalón, ventanas |

Un número complejo $z = a + jb$ también se escribe en forma polar $z = |z|\,e^{j\theta}$, con

$$|z| = \sqrt{a^2+b^2}, \qquad \theta = \arctan\!\left(\frac{b}{a}\right)\ \text{(en el cuadrante correcto)}.$$

La **fórmula de Euler**, $e^{j\theta} = \cos\theta + j\,\sin\theta$, será la base de todo Fourier.

### Ejemplos

```matlab
% Ejemplo 1.2.1: tipos de datos y la función class()
x = 3.7;            % double: real de doble precisión (tipo por defecto)
c = 'Señal';        % char: arreglo de caracteres (comillas simples)
b = (x > 2);        % logical: 1 (verdadero) o 0 (falso)
class(x)
class(c)
class(b)
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `x = 3.7;`: un número real. MATLAB lo guarda como `double` (doble precisión, unas 16 cifras significativas), el tipo por defecto.
- `c = 'Señal';`: texto entre **comillas simples** = arreglo de caracteres (`char`). Cada letra ocupa una posición.
- `b = (x > 2);`: la comparación `x > 2` es verdadera, así que `b` vale `1` y es de tipo `logical` (verdadero/falso).
- `class(x)`, `class(c)`, `class(b)`: `class` devuelve el tipo de dato de la variable: `'double'`, `'char'` y `'logical'`.

</details>

```matlab
% Ejemplo 1.2.2: números complejos (i o j son la unidad imaginaria)
z = 3 + 4j;
parte_real = real(z)        % 3
parte_imag = imag(z)        % 4
magnitud   = abs(z)         % |z| = sqrt(3^2 + 4^2) = 5
fase_rad   = angle(z)       % ángulo en radianes (usa el cuadrante correcto)
fase_grad  = angle(z)*180/pi
z_conj     = conj(z)        % conjugado: 3 - 4j
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `z = 3 + 4j;`: crea el número complejo $3+4j$. La `j` (o `i`) pegada al número indica la unidad imaginaria $\sqrt{-1}$.
- `real(z)`: devuelve la parte real, 3.
- `imag(z)`: devuelve la parte imaginaria, 4 (sin la `j`).
- `abs(z)`: devuelve la magnitud $|z| = \sqrt{3^2+4^2} = 5$. Con números reales, `abs` es el valor absoluto.
- `angle(z)`: devuelve la fase en **radianes**, calculada en el cuadrante correcto (≈ 0.9273 rad).
- `angle(z)*180/pi`: convierte radianes a grados multiplicando por $180/\pi$ (≈ 53.13°). `pi` es la constante π ya definida en MATLAB.
- `conj(z)`: devuelve el conjugado, $3-4j$ (cambia el signo de la parte imaginaria).

</details>

```matlab
% Ejemplo 1.2.3: forma polar y fórmula de Euler
z2 = 5*exp(1j*pi/3)          % |z|=5, ángulo 60°  ->  2.5 + 4.33j
euler = exp(1j*pi) + 1       % identidad de Euler: debe dar ~0 (error de redondeo)
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `z2 = 5*exp(1j*pi/3)`: construye un complejo en **forma polar** $|z|e^{j\theta}$.
  - `exp(...)`: la función exponencial $e^{(\cdot)}$; acepta argumentos complejos.
  - `1j*pi/3`: el exponente $j\pi/3$, es decir, un ángulo de 60°. Se escribe `1j` (y no solo `j`) para que no se confunda con una variable llamada `j`.
  - El resultado se muestra en forma rectangular: $2.5 + 4.33j$.
- `euler = exp(1j*pi) + 1`: verifica la identidad de Euler $e^{j\pi}+1=0$. Da un número muy pequeño (≈ $10^{-16}j$) y no exactamente 0 por el **error de redondeo** de la computadora.

</details>

```matlab
% Ejemplo 1.2.4: constantes especiales y formato de visualización
format long
pi                % 3.141592653589793
format short      % volvemos al formato normal (4 decimales)
pi
infinito = 1/0    % Inf
indeterm = 0/0    % NaN (Not a Number)
precision = eps   % el número más pequeño tal que 1 + eps > 1
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `format long`: a partir de aquí MATLAB muestra 15 decimales.
- `pi`: muestra π con todos esos decimales.
- `format short`: vuelve al formato normal de 4 decimales (solo cambia cómo se **muestra**, no la precisión del cálculo).
- `infinito = 1/0`: dividir entre cero no da error en MATLAB; da `Inf` (infinito).
- `indeterm = 0/0`: una indeterminación da `NaN` (*Not a Number*).
- `precision = eps`: `eps` es la separación entre 1 y el siguiente número representable (≈ $2.2\times10^{-16}$). Sirve para saber cuándo una diferencia es solo error de redondeo.

</details>

## 1.3 Operadores

### Concepto

| Tipo | Operadores | Nota clave |
|---|---|---|
| Aritméticos (matriciales) | `+  -  *  /  ^` | `*` es producto **matricial** (filas × columnas) |
| Elemento a elemento | `.*  ./  .^` | Opera posición por posición, que es lo que necesitas con señales |
| Relacionales | `==  ~=  <  >  <=  >=` | Devuelven `logical` (0/1) |
| Lógicos | `&  \|  ~` (y `&&`, `\|\|` para escalares) | Combinan condiciones |

> 💡 **Analogía:** tienes dos listas de mercado, *precios* y *cantidades*. Para saber cuánto pagas por cada producto multiplicas **posición por posición** (`.*`). El producto matricial (`*`) es otra operación, con otras reglas.

**Regla práctica:** si `t` es un vector de tiempo, toda expresión que dependa de `t` debe usar `.*`, `./` y `.^`.

### Ejemplos

```matlab
% Ejemplo 1.3.1: producto matricial vs. elemento a elemento
A = [1 2; 3 4];
B = [5 6; 7 8];
A*B       % producto matricial: [19 22; 43 50]
A.*B      % producto elemento a elemento: [5 12; 21 32]
A^2       % A*A (matricial)
A.^2      % cada elemento al cuadrado
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `A = [1 2; 3 4];`: crea una matriz 2×2. Dentro de los corchetes, los **espacios** (o comas) separan columnas y el **punto y coma** separa filas.
- `B = [5 6; 7 8];`: otra matriz 2×2.
- `A*B`: producto **matricial** (fila por columna, como en álgebra lineal): $[1\cdot5+2\cdot7,\ \dots]$.
- `A.*B`: producto **elemento a elemento**: multiplica cada posición con la misma posición de la otra matriz. El punto antes del operador es lo que cambia el significado.
- `A^2`: potencia matricial, igual a `A*A`.
- `A.^2`: eleva al cuadrado **cada elemento** por separado.

</details>

```matlab
% Ejemplo 1.3.2: operar sobre un vector de tiempo
t = 0:0.25:1;              % vector de tiempo
y = t.^2 .* exp(-t)        % CORRECTO: elemento a elemento
% y = t^2 * exp(-t)        % ERROR: t^2 intenta t*t (matricial) con un vector fila
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `t = 0:0.25:1;`: crea el vector `[0 0.25 0.5 0.75 1]` (la sintaxis `inicio:paso:fin` se explica en detalle en 1.4).
- `y = t.^2 .* exp(-t)`: evalúa $y = t^2 e^{-t}$ en cada instante.
  - `t.^2`: cada tiempo al cuadrado.
  - `exp(-t)`: la exponencial de cada elemento.
  - `.*`: multiplica las dos listas posición por posición. Si se usara `*`, MATLAB intentaría un producto matricial entre dos vectores fila y daría error.
- `% y = t^2 * exp(-t)`: línea comentada que muestra la forma **incorrecta**. Quita el `%` para ver el error que produce.

</details>

```matlab
% Ejemplo 1.3.3: operadores relacionales -> ¡la semilla del escalón unitario!
t = -2:0.5:2;
u = (t >= 0)               % 1 donde t >= 0 y 0 donde t < 0 (tipo logical)
u = double(u)              % lo convertimos a double para operar con él
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `t = -2:0.5:2;`: vector de tiempo de −2 a 2 con paso 0.5 (9 valores).
- `u = (t >= 0)`: compara **cada** elemento de `t` con 0. Donde la condición se cumple, pone `1`; donde no, `0`. El resultado es un vector `logical` del mismo tamaño que `t`. ¡Esto ya es el escalón unitario $u(t)$!
- `u = double(u)`: `double` convierte el vector lógico a números `double`, para poder sumarlo, multiplicarlo o graficarlo sin problemas.

</details>

```matlab
% Ejemplo 1.3.4: operadores lógicos para construir una "ventana"
t = -2:0.5:2;
ventana = (t >= -1) & (t <= 1)    % 1 solo dentro del intervalo [-1, 1]
fuera   = ~ventana                % negación: 1 fuera del intervalo
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `ventana = (t >= -1) & (t <= 1)`: dos comparaciones unidas con `&` (Y lógico). El resultado vale 1 solo donde **ambas** son verdaderas, es decir, para $-1 \le t \le 1$.
- `fuera = ~ventana`: `~` es la negación (NO lógico): cambia los 1 por 0 y los 0 por 1.
- Existe también `|` (O lógico): vale 1 si **al menos una** condición es verdadera.

</details>

## 1.4 Vectores y matrices

### Concepto

| Instrucción | Resultado |
|---|---|
| `[1 2 3]` / `[1; 2; 3]` | Vector fila / vector columna |
| `a:paso:b` | De `a` a `b` con incremento `paso` |
| `linspace(a, b, N)` | `N` puntos equiespaciados entre `a` y `b` |
| `v(k)`, `v(end)`, `v(2:4)` | Indexación: **MATLAB empieza en 1**, no en 0 |
| `zeros`, `ones`, `eye` | Matrices de ceros, unos, identidad |
| `size`, `length`, `numel` | Dimensiones, longitud mayor, número de elementos |
| `'` / `.'` | Transpuesta conjugada / transpuesta simple |

En señales, el vector de tiempo se construye a partir de la **frecuencia de muestreo** $f_s$:

$$T_s = \frac{1}{f_s}, \qquad t = 0,\ T_s,\ 2T_s,\ \dots$$

### Ejemplos

```matlab
% Ejemplo 1.4.1: formas de crear vectores
fila    = [1 2 3 4]           % vector fila (1x4)
columna = [1; 2; 3; 4]        % vector columna (4x1)
v1 = 0:0.2:1                  % operador dos puntos: inicio:paso:fin
v2 = linspace(0, 1, 6)        % 6 puntos entre 0 y 1 (el mismo resultado)
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `fila = [1 2 3 4]`: vector **fila** (1×4); los espacios separan columnas.
- `columna = [1; 2; 3; 4]`: vector **columna** (4×1); los `;` separan filas.
- `v1 = 0:0.2:1`: operador dos puntos con la sintaxis `inicio : paso : fin`.
  - `0` (inicio): primer valor.
  - `0.2` (paso): incremento entre valores consecutivos.
  - `1` (fin): último valor permitido. Si el paso no "cae" exactamente en el fin, MATLAB se detiene antes de pasarse.
  - Resultado: `[0 0.2 0.4 0.6 0.8 1]`. Si se omite el paso (`1:5`), se asume 1.
- `v2 = linspace(0, 1, 6)`: crea **6 puntos** igualmente espaciados entre 0 y 1, incluyendo ambos extremos.
  - Parámetros: `linspace(inicio, fin, número_de_puntos)`.
  - Diferencia clave: con `:` eliges el **paso**; con `linspace` eliges el **número de puntos**.

</details>

```matlab
% Ejemplo 1.4.2: indexación
v = 10:10:100;        % [10 20 ... 100]
v(3)                  % tercer elemento -> 30  (¡el primer índice es 1!)
v(end)                % último elemento -> 100
v(2:4)                % elementos 2, 3 y 4
v([1 end])            % primero y último
v(v > 50)             % indexación lógica: los elementos mayores que 50
v(4) = 0              % modificar un elemento
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `v = 10:10:100;`: vector `[10 20 30 ... 100]` (10 elementos).
- `v(3)`: el **paréntesis** accede a la posición 3, que es 30. En MATLAB **la primera posición es 1**, no 0.
- `v(end)`: `end` dentro del paréntesis significa "la última posición".
- `v(2:4)`: un vector de índices extrae varias posiciones a la vez: 20, 30, 40.
- `v([1 end])`: extrae posiciones salteadas, la primera y la última.
- `v(v > 50)`: **indexación lógica**. `v > 50` crea un vector de 0 y 1, y al usarlo como índice se extraen solo los elementos donde hay 1.
- `v(4) = 0`: asigna un nuevo valor a la posición 4.

</details>

```matlab
% Ejemplo 1.4.3: matrices especiales y dimensiones
Z = zeros(2,3)        % matriz 2x3 de ceros
O = ones(1,4)         % vector de unos
I3 = eye(3)           % identidad 3x3
M = [1 2 3; 4 5 6];
[filas, cols] = size(M)   % 2 filas, 3 columnas
numel(M)                  % 6 elementos en total
M(2,3)                    % fila 2, columna 3 -> 6
M(:,2)                    % toda la columna 2
```

<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `ones(1,4)`: vector fila de cuatro unos.
- `eye(3)`: matriz identidad 3×3 (unos en la diagonal).
- `M = [1 2 3; 4 5 6];`: matriz de 2 filas y 3 columnas.
- `[filas, cols] = size(M)`: `size` devuelve **dos** salidas, número de filas y de columnas. Los corchetes a la izquierda reciben cada una.
- `numel(M)`: número total de elementos (6). `length` devolvería la dimensión mayor (3).
- `M(2,3)`: elemento de la fila 2, columna 3. Con matrices se usa `(fila, columna)`.
- `M(:,2)`: los dos puntos solos significan "todas". Aquí: todas las filas de la columna 2.

</details>

## 1.5 Funciones básicas

### Concepto

| Grupo | Funciones |
|---|---|
| Elementales | `sin  cos  tan  exp  log  log10  sqrt  abs` |
| Estadísticas | `sum  mean  max  min  std  sort` |
| Búsqueda | `find`, `any`, `all` |
| Redondeo y residuo | `round  floor  ceil  fix  mod  rem` |
| Acumulación y diferencias | `cumsum` (≈ integral), `diff` (≈ derivada) |

Todas las funciones elementales **operan sobre vectores completos**: `sin(t)` calcula el seno de cada muestra.

En tiempo discreto, `diff` y `cumsum` aproximan la derivada y la integral:

$$\frac{dx}{dt} \approx \frac{x[n+1]-x[n]}{T_s}, \qquad \int x\,dt \approx T_s\sum_{k} x[k].$$

### Ejemplos

```matlab
% Ejemplo 1.5.1: funciones elementales sobre vectores
t = [0 pi/6 pi/4 pi/2];
s = sin(t)            % seno de cada elemento
e = exp(-t)           % exponencial decreciente
r = sqrt([4 9 16])    % raíz cuadrada elemento a elemento
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `t = [0 pi/6 pi/4 pi/2];`: cuatro ángulos en radianes.
- `s = sin(t)`: `sin` calcula el seno de **cada** elemento y devuelve un vector del mismo tamaño. Las funciones trigonométricas trabajan en **radianes** (para grados existe `sind`).
- `e = exp(-t)`: $e^{-t}$ para cada elemento.
- `r = sqrt([4 9 16])`: raíz cuadrada de cada elemento: `[2 3 4]`.

</details>

```matlab
% Ejemplo 1.5.2: estadística sobre las lecturas de un sensor
lect = [2.1 2.4 2.2 3.8 2.3 2.5 2.2];   % lecturas de un sensor [V]
total    = sum(lect)
promedio = mean(lect)
desv     = std(lect)
[vmax, imax] = max(lect)     % valor máximo y en qué posición ocurre
valor_rms = sqrt(mean(lect.^2))   % valor eficaz (RMS)
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `lect = [2.1 2.4 ... 2.2];`: siete lecturas de un sensor en voltios.
- `sum(lect)`: suma todos los elementos.
- `mean(lect)`: promedio (suma dividida entre el número de elementos).
- `std(lect)`: desviación estándar: qué tanto se dispersan las lecturas alrededor del promedio.
- `[vmax, imax] = max(lect)`: `max` puede devolver **dos** salidas.
  - `vmax`: el valor máximo (3.8).
  - `imax`: la **posición** donde ocurre (4).
  - `min` funciona igual para el mínimo.
- `sqrt(mean(lect.^2))`: valor eficaz o RMS, $\sqrt{\frac{1}{N}\sum x^2}$: eleva al cuadrado, promedia y saca raíz.

</details>

```matlab
% Ejemplo 1.5.3: find para detectar cuándo una señal supera un umbral
umbral = 2.4;
idx = find(lect > umbral)          % posiciones donde se supera el umbral
primera = find(lect > umbral, 1)   % solo la primera vez
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `umbral = 2.4;`: nivel de referencia.
- `idx = find(lect > umbral)`: `lect > umbral` produce 0 y 1; `find` devuelve las **posiciones** donde hay 1 (es decir, dónde se supera el umbral).
- `find(lect > umbral, 1)`: el segundo parámetro limita el resultado a las **primeras k** posiciones; con 1, devuelve solo la primera vez que se supera el umbral.

</details>

```matlab
% Ejemplo 1.5.4: redondeo y residuo; mod es ideal para fenómenos periódicos
round(2.5), floor(2.7), ceil(2.1), fix(-2.7)
angulo = 0:90:720;            % un eje que gira dos vueltas [grados]
angulo_encoder = mod(angulo, 360)   % el encoder absoluto "reinicia" cada 360°
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `round(2.5)`: redondea al entero más cercano (3). `floor(2.7)`: redondea hacia abajo (2). `ceil(2.1)`: hacia arriba (3). `fix(-2.7)`: hacia el cero (−2).
- `angulo = 0:90:720;`: ángulos de 0° a 720° (dos vueltas) cada 90°.
- `mod(angulo, 360)`: `mod(a, m)` devuelve el **residuo** de dividir `a` entre `m`. Así 450° se convierte en 90°: el valor "da la vuelta" cada 360, como un encoder absoluto o un reloj.

</details>

```matlab
% Ejemplo 1.5.5: diff y cumsum = derivada e integral discretas
Ts  = 0.1;                       % periodo de muestreo [s]
pos = [0 0.5 2 4.5 8 12.5];      % posición de un carro [m]
vel = diff(pos)/Ts               % velocidad aproximada [m/s] (un elemento menos)
pos_rec = [0 cumsum(vel)*Ts]     % integrar la velocidad devuelve la posición
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `Ts = 0.1;`: tiempo entre muestras.
- `pos = [0 0.5 2 4.5 8 12.5];`: posiciones medidas.
- `vel = diff(pos)/Ts`: `diff` calcula las diferencias entre elementos consecutivos, `[pos(2)-pos(1), pos(3)-pos(2), ...]`. Al dividir entre `Ts` se obtiene la velocidad aproximada $\Delta x/\Delta t$. El resultado tiene **un elemento menos** que `pos`.
- `pos_rec = [0 cumsum(vel)*Ts]`: `cumsum` hace la **suma acumulada** `[v1, v1+v2, v1+v2+v3, ...]`. Multiplicada por `Ts` aproxima la integral. Se antepone el 0 inicial con los corchetes para recuperar la posición original.

</details>

## 1.6 Control de flujo

### Concepto

```matlab
if condicion            for k = 1:N              while condicion          switch variable
    ...                     ...                      ...                      case 'a'
elseif otra                 end                      end                          ...
    ...                                                                       otherwise
else                                                                              ...
    ...                                                                   end
end
```

> 💡 **Vectorizar:** en MATLAB casi siempre es mejor operar con vectores completos que recorrerlos con un `for`. El código es más corto y mucho más rápido. Aun así, los bucles son indispensables para **sistemas recursivos**, como $y[n] = a\,y[n-1] + x[n]$.

### Ejemplos

```matlab
% Ejemplo 1.6.1: if / elseif / else -> clasificar la temperatura de un motor
T_motor = 78;      % [°C]
if T_motor < 60
    disp('Operación normal')
elseif T_motor < 85
    disp('Advertencia: temperatura elevada')
else
    disp('ALARMA: detener el motor')
end
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `T_motor = 78;`: temperatura medida.
- `if T_motor < 60`: si la condición es verdadera, ejecuta las líneas siguientes hasta el próximo `elseif`, `else` o `end`.
- `disp('Operación normal')`: muestra el mensaje.
- `elseif T_motor < 85`: se evalúa **solo si** la condición anterior fue falsa.
- `else`: se ejecuta si ninguna condición anterior fue verdadera.
- `end`: cierra el bloque `if`. Todo `if`, `for`, `while` y `switch` termina con `end`.

</details>

```matlab
% Ejemplo 1.6.2: for -> suma parcial de la serie geométrica sum(0.5^k)
S = 0;
for k = 0:10
    S = S + 0.5^k;     % acumulamos cada término
end
S                       % se acerca a 1/(1-0.5) = 2
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `S = 0;`: inicializa el acumulador en cero.
- `for k = 0:10`: repite el bloque **11 veces**; en cada vuelta `k` toma el siguiente valor del vector `0:10`.
- `S = S + 0.5^k;`: suma el término $0.5^k$ al acumulador (el nuevo `S` es el viejo `S` más el término).
- `end`: cierra el bucle.
- `S`: muestra el resultado, 1.999, cercano al valor exacto de la serie infinita, 2.

</details>

```matlab
% Ejemplo 1.6.3: while -> ¿cuánto tarda e^(-t/tau) en caer por debajo del 5 %?
tau = 0.2;  Ts = 0.001;  t = 0;
while exp(-t/tau) > 0.05
    t = t + Ts;         % avanzamos en el tiempo hasta cumplir la condición
end
t_5porciento = t        % ~ 3*tau = 0.6 s (criterio del 5 %)
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `tau = 0.2; Ts = 0.001; t = 0;`: constante de tiempo, paso de tiempo y tiempo inicial (tres asignaciones en una línea).
- `while exp(-t/tau) > 0.05`: repite el bloque **mientras** la exponencial siga por encima del 5 %. No se sabe de antemano cuántas vueltas se necesitan; por eso se usa `while` y no `for`.
- `t = t + Ts;`: avanza el tiempo un paso. Si olvidas actualizar la variable de la condición, el bucle nunca termina (en ese caso detén MATLAB con **Ctrl + C**).
- `t_5porciento = t`: el instante en que se cumplió la condición, ≈ 0.6 s = $3\tau$.

</details>

```matlab
% Ejemplo 1.6.4: switch -> elegir el tipo de señal
tipo = 'cuadrada';
t = 0:0.25:1;
switch tipo
    case 'seno'
        x = sin(2*pi*t);
    case 'cuadrada'
        x = double(mod(t, 1) < 0.5);
    otherwise
        x = zeros(size(t));
end
x
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `tipo = 'cuadrada';`: texto que decide qué señal generar.
- `switch tipo`: compara `tipo` con cada `case`.
- `case 'seno'`: si `tipo` es `'seno'`, ejecuta la línea siguiente.
- `case 'cuadrada'`: `x = double(mod(t, 1) < 0.5);` crea una onda cuadrada: vale 1 durante la primera mitad de cada segundo.
- `otherwise`: se ejecuta si ningún `case` coincidió; aquí `zeros(size(t))` crea ceros del mismo tamaño que `t`.
- `end`: cierra el `switch`.

</details>

```matlab
% Ejemplo 1.6.5: bucle vs. vectorización (compara los tiempos)
t = linspace(0, 1, 1e6);
tic
y1 = zeros(size(t));
for k = 1:numel(t)
    y1(k) = sin(2*pi*5*t(k));     % muestra por muestra
end
t_bucle = toc
tic
y2 = sin(2*pi*5*t);               % todo el vector de una vez
t_vector = toc
iguales = max(abs(y1 - y2))       % el mismo resultado
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `t = linspace(0, 1, 1e6);`: un millón de puntos (`1e6` = $1\times10^6$).
- `tic`: arranca un cronómetro.
- `y1 = zeros(size(t));`: reserva memoria para el resultado (hace el bucle más rápido).
- `for k = 1:numel(t)` … `y1(k) = sin(2*pi*5*t(k));`: calcula el seno **muestra por muestra**, un millón de vueltas.
- `t_bucle = toc`: `toc` detiene el cronómetro y devuelve los segundos transcurridos desde `tic`.
- `y2 = sin(2*pi*5*t);`: el mismo cálculo **vectorizado**, todo el vector de una vez.
- `iguales = max(abs(y1 - y2))`: la mayor diferencia entre ambos resultados es 0: son idénticos, pero la versión vectorizada es más rápida.

</details>

## 1.7 Scripts y funciones

### Concepto

| Herramienta | Qué es | Cuándo usarla |
|---|---|---|
| **Script** (`.m`) | Secuencia de instrucciones que comparte el workspace | Prácticas, experimentos |
| **Función anónima** | `f = @(t) expresión` en una sola línea | Definir señales matemáticas, como $x(t)=e^{-2t}\cos(6\pi t)$ |
| **Archivo de función** | `function [salidas] = nombre(entradas)` en su propio `.m` | Cálculos reutilizables con varias líneas |

En el Módulo 2 las funciones anónimas serán nuestra herramienta principal: una señal **es** una función del tiempo, y con ellas se escribe tal cual, por ejemplo `x = @(t) ...`.

### Ejemplos

```matlab
% Ejemplo 1.7.1: función anónima de una variable
f = @(t) exp(-2*t) .* cos(2*pi*3*t);   % señal amortiguada
f(0)                 % evaluar en un punto -> 1
f([0 0.1 0.2])       % evaluar en un vector (gracias a .*)
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `f = @(t) exp(-2*t) .* cos(2*pi*3*t);`: crea una **función anónima**.
  - `@`: indica que se está creando una función.
  - `(t)`: el nombre de la variable de entrada.
  - Lo que sigue es la fórmula: $f(t)=e^{-2t}\cos(6\pi t)$. Se usa `.*` para que funcione también con vectores.
  - `f` es una variable que guarda la **función**, no un número.
- `f(0)`: evalúa la función en $t = 0$: $e^0\cos 0 = 1$.
- `f([0 0.1 0.2])`: evalúa la función en tres instantes a la vez y devuelve tres valores.

</details>

```matlab
% Ejemplo 1.7.2: función anónima con varios parámetros
senoidal = @(A, f0, phi, t) A*sin(2*pi*f0*t + phi);
t = 0:0.05:0.2;
x1 = senoidal(2, 5, 0, t)          % amplitud 2, 5 Hz, sin desfase
x2 = senoidal(1, 5, pi/2, t)       % desfase de 90°
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `senoidal = @(A, f0, phi, t) A*sin(2*pi*f0*t + phi);`: función anónima con **cuatro** entradas, separadas por comas: amplitud `A`, frecuencia `f0` en Hz, fase `phi` en radianes y tiempo `t`.
- `t = 0:0.05:0.2;`: cinco instantes de tiempo.
- `x1 = senoidal(2, 5, 0, t)`: los valores se asignan **en el mismo orden** de la definición: `A = 2`, `f0 = 5`, `phi = 0`, `t = t`.
- `x2 = senoidal(1, 5, pi/2, t)`: amplitud 1 y fase de 90° ($\pi/2$), que convierte el seno en coseno.

</details>

**Ejemplo 1.7.3: archivo de función.** En MATLAB Online usa *New → Function*, pega el código y guárdalo como `energia_potencia.m`:

```matlab
% Ejemplo 1.7.3 (uso): llamar a la función con dos salidas
x = [1 2 3 2 1];
[E, P] = energia_potencia(x)     % E = 1+4+9+4+1 = 19,  P = 19/5 = 3.8
```

## 2.1 Señales continuas y discretas

### Concepto
- **Señal continua (CT):** $x(t)$, definida para todo $t\in\mathbb{R}$. Ejemplo: el voltaje de un tacogenerador.
- **Señal discreta (DT):** $x[n]$, definida solo para $n\in\mathbb{Z}$. Ejemplo: las lecturas del ADC de un microcontrolador.

El vínculo entre ambas es el **muestreo** con periodo $T_s$:

$$x[n] = x(nT_s), \qquad f_s = \frac{1}{T_s}.$$

> 💡 **Analogía:** una película (continua) frente a las **fotos** tomadas cada cierto tiempo (discreta). Si tomas pocas fotos por segundo, se pierde el movimiento rápido.

### Ejemplos

```matlab
% Ejemplo 2.1.1: señal "continua" -> vector de tiempo muy denso + plot
t = 0:0.001:1;                 % 1001 puntos: paso pequeño => apariencia continua
x = cos(2*pi*2*t);             % x(t) = cos(4*pi*t), f = 2 Hz
figure
plot(t, x, 'Color', navy); grid on
xlabel('t [s]'); ylabel('x(t)'); title('Señal continua: x(t) = cos(4\pi t)')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `t = 0:0.001:1;`: vector de tiempo con la sintaxis `inicio : paso : fin`.
  - `0` (inicio): el tiempo empieza en 0.
  - `0.001` (paso): avanza de 1 ms en 1 ms ($10^{-3}$ s): `0, 0.001, 0.002, ...`.
  - `1` (fin): termina en 1 s. En total son 1001 puntos.
  - `;`: evita que MATLAB imprima los 1001 números.
  - El paso tan pequeño hace que, al unir los puntos, la curva **parezca continua**.
- `x = cos(2*pi*2*t);`: evalúa $x(t)=\cos(2\pi f t)$ con $f = 2$ Hz en cada uno de los 1001 instantes. `x` también tiene 1001 elementos.
- `figure`: abre una ventana nueva para la gráfica.
- `plot(t, x, 'Color', navy); grid on`:
  - `t` → eje horizontal (tiempo); `x` → eje vertical (amplitud).
  - `'Color', navy`: color azul marino definido en la configuración.
  - `grid on`: rejilla de fondo.
- `xlabel`, `ylabel`, `title`: etiquetas y título. En el título, `\pi` se muestra como π.

</details>

```matlab
% Ejemplo 2.1.2: señal discreta -> índice entero n + stem
n = 0:15;                      % índice entero
xn = cos(2*pi*n/8);            % x[n] = cos(pi*n/4), periodo N = 8
figure
stem(n, xn, 'filled', 'Color', cian); grid on
xlabel('n'); ylabel('x[n]'); title('Señal discreta: x[n] = cos(\pi n/4)')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `n = 0:15;`: índice **entero** de 0 a 15 (paso 1 por defecto). En tiempo discreto la variable independiente es `n`, no `t`.
- `xn = cos(2*pi*n/8);`: $x[n] = \cos(2\pi n/8)$. Al ser $\Omega_0 = \pi/4$, se repite cada 8 muestras.
- `stem(n, xn, 'filled', 'Color', cian)`: gráfica de **bastones**, la forma correcta de mostrar una señal discreta.
  - `n`: posiciones en el eje horizontal.
  - `xn`: altura de cada bastón.
  - `'filled'`: rellena los círculos de la punta.
  - `'Color', cian`: color de los bastones.

</details>

```matlab
% Ejemplo 2.1.3: muestreo -> superponer x(t) y sus muestras x[n] = x(nTs)
Ts = 0.05;                     % periodo de muestreo [s]  (fs = 20 Hz)
n  = 0:round(1/Ts);            % índices que cubren 1 s
tn = n*Ts;                     % instantes de muestreo
figure
plot(t, x, 'Color', navy); hold on
stem(tn, cos(2*pi*2*tn), 'filled', 'Color', ambar)
hold off; grid on
xlabel('t [s]'); legend('x(t)', 'x[n] = x(nT_s)')
title('Muestreo con f_s = 20 Hz')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `Ts = 0.05;`: periodo de muestreo de 50 ms, es decir, $f_s = 1/T_s = 20$ Hz.
- `n = 0:round(1/Ts);`: índices de 0 a 20. `round` asegura un número entero (1/0.05 = 20).
- `tn = n*Ts;`: convierte cada índice en su instante real: $t_n = nT_s$.
- `plot(t, x, 'Color', navy); hold on`: dibuja la señal "continua" del Ejemplo 2.1.1 y congela la gráfica.
- `stem(tn, cos(2*pi*2*tn), 'filled', 'Color', ambar)`: dibuja las muestras $x[n] = x(nT_s)$ encima, en sus instantes reales.
- `legend('x(t)', 'x[n] = x(nT_s)')`: `_s` produce el subíndice.

</details>

## 2.2 `plot`, `stem` y `stairs`

### Concepto

| Función | Representa | Uso típico |
|---|---|---|
| `plot` | Une los puntos con rectas | Señales CT (vector denso) |
| `stem` | Un "palito" por muestra | Señales DT $x[n]$ |
| `stairs` | Mantiene cada valor hasta la siguiente muestra | Salida de un DAC o retenedor de orden cero (ZOH) |

> ⚠️ Usar `plot` con una señal discreta **inventa** valores entre muestras que no existen.

### Ejemplos

```matlab
% Ejemplo 2.2.1: la misma secuencia con las tres representaciones
n  = 0:10;
xn = [0 2 3 1 -1 -2 0 1 2 1 0];
figure
subplot(3,1,1); plot(n, xn, 'o-', 'Color', navy); grid on
title('plot: une muestras (engañoso para DT)')
subplot(3,1,2); stem(n, xn, 'filled', 'Color', cian); grid on
title('stem: la representación correcta de x[n]')
subplot(3,1,3); stairs(n, xn, 'Color', ambar); grid on
title('stairs: lo que entrega un DAC (retenedor de orden cero)'); xlabel('n')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `subplot(3,1,k)`: tres gráficas apiladas (3 filas, 1 columna); `k` elige cuál.
- `plot(n, xn, 'o-', 'Color', navy)`: `'o-'` dibuja círculos en las muestras **y** líneas que los unen. Las líneas inventan valores entre muestras que no existen.
- `stem(n, xn, 'filled', 'Color', cian)`: representación correcta, un bastón por muestra.
- `stairs(n, xn, 'Color', ambar)`: dibujo en **escalera**: mantiene cada valor constante hasta la siguiente muestra, como la salida de un convertidor digital-analógico (DAC).

</details>

## 2.3 Señales elementales

### Concepto

| Señal | Tiempo continuo | Tiempo discreto |
|---|---|---|
| Escalón unitario | $u(t)=\begin{cases}1 & t\ge 0\\0 & t<0\end{cases}$ | $u[n]=\begin{cases}1 & n\ge 0\\0 & n<0\end{cases}$ |
| Impulso unitario | $\delta(t)$: área 1, ancho → 0 | $\delta[n]=\begin{cases}1 & n=0\\0 & n\ne 0\end{cases}$ |
| Rampa | $r(t)=t\,u(t)$ | $r[n]=n\,u[n]$ |
| Pulso rectangular | $\Pi(t)=u(t+\tfrac12)-u(t-\tfrac12)$ | |
| Exponencial real | $x(t)=Ce^{at}$ | $x[n]=C\alpha^n$ |
| Senoidal | $x(t)=A\cos(\omega_0 t+\phi)$ | $x[n]=A\cos(\Omega_0 n+\phi)$ |
| Exponencial compleja | $x(t)=e^{(\sigma+j\omega_0)t}$ | $x[n]=e^{j\Omega_0 n}$ |
| Sinc | $\mathrm{sinc}(t)=\dfrac{\sin(\pi t)}{\pi t}$ | |

Relaciones clave: $u(t)=\int_{-\infty}^{t}\delta(\tau)d\tau$ y $r(t)=\int_{-\infty}^{t}u(\tau)d\tau$.

### Ejemplos

```matlab
% Ejemplo 2.3.1: escalón unitario en CT y DT
t = -2:0.001:3;
n = -5:10;
figure
subplot(2,1,1); plot(t, double(t >= 0), 'Color', navy); grid on
axis([-2 3 -0.2 1.2]); title('u(t)'); xlabel('t')
subplot(2,1,2); stem(n, double(n >= 0), 'filled', 'Color', cian); grid on
axis([-5 10 -0.2 1.2]); title('u[n]'); xlabel('n')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `t = -2:0.001:3;` y `n = -5:10;`: ejes de tiempo continuo (denso) y discreto (entero).
- `double(t >= 0)`: la comparación da 1 donde $t \ge 0$ y 0 donde no; `double` la convierte a número. Es exactamente $u(t)$.
- `axis([-2 3 -0.2 1.2])`: deja un pequeño margen arriba y abajo para que el escalón no quede pegado al borde.
- `stem(n, double(n >= 0), 'filled', 'Color', cian)`: la misma idea con el índice entero: $u[n]$.

</details>

```matlab
% Ejemplo 2.3.2: impulso discreto y la aproximación del impulso continuo
n = -5:5;
figure
subplot(1,2,1)
stem(n, double(n == 0), 'filled', 'Color', cian); grid on
title('\delta[n]'); xlabel('n'); axis([-5 5 -0.2 1.2])
subplot(1,2,2); hold on
anchos = [0.5 0.2 0.1];  colores = {navy, cian, ambar};
t = -1:0.001:1;
for k = 1:3
    D = anchos(k);
    pulso = (1/D) * double(abs(t) <= D/2);   % ancho D, altura 1/D => área = 1
    plot(t, pulso, 'Color', colores{k})
    area_k = trapz(t, pulso);                 % verificación numérica del área
    fprintf('Ancho %.1f -> área = %.3f\n', D, area_k)
end
hold off; grid on; xlabel('t')
title('\delta(t) como límite de pulsos de área 1')
legend('\Delta = 0.5', '\Delta = 0.2', '\Delta = 0.1')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `stem(n, double(n == 0), ...)`: `n == 0` vale 1 solo en $n = 0$; es el impulso discreto $\delta[n]$. Ojo: `==` compara, `=` asigna.
- `subplot(1,2,1)` / `subplot(1,2,2)`: dos gráficas lado a lado (1 fila, 2 columnas).
- `anchos = [0.5 0.2 0.1];`: tres anchos de pulso.
- `colores = {navy, cian, ambar};`: las **llaves** `{}` crean un *cell array*, una lista que puede guardar elementos de cualquier tipo (aquí, tres vectores RGB). Se accede con `colores{k}`.
- `for k = 1:3`: repite para cada ancho.
- `pulso = (1/D) * double(abs(t) <= D/2);`: pulso centrado de ancho `D` y altura `1/D`, para que el área sea siempre 1.
- `area_k = trapz(t, pulso);`: `trapz(x, y)` calcula la **integral numérica** (área bajo la curva) por el método del trapecio.
- `fprintf('Ancho %.1f -> área = %.3f\n', D, area_k)`: imprime texto con formato: `%.1f` y `%.3f` se reemplazan por los valores con 1 y 3 decimales; `\n` es salto de línea.

</details>

```matlab
% Ejemplo 2.3.3: rampa y pulso rectangular
t = -2:0.001:3;
figure
subplot(2,1,1); plot(t, t.*(t >= 0), 'Color', navy); grid on
title('r(t) = t u(t)'); xlabel('t')
subplot(2,1,2); plot(t, double(abs(t) <= 0.5), 'Color', cian); grid on
axis([-2 3 -0.2 1.2]); title('\Pi(t): pulso unitario de ancho 1'); xlabel('t')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `plot(t, t.*(t >= 0), ...)`: la rampa $r(t) = t\,u(t)$: multiplica el tiempo por el escalón, así que vale 0 para $t<0$ y $t$ para $t \ge 0$.
- `plot(t, double(abs(t) <= 0.5), ...)`: el pulso rectangular $\Pi(t)$: vale 1 donde $|t| \le 0.5$. `abs` es el valor absoluto.

</details>

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

</details>

- `plot(t, A*cos(w0*t), ...)`: fase $\phi = 0$.
- `plot(t, A*cos(w0*t - pi/2), '--', ...)`: fase $-\pi/2$: la curva se **retrasa** un cuarto de periodo y se convierte en seno.
- `plot(t, A*cos(w0*t + pi), ':', ...)`: fase $\pi$: la señal queda **invertida**.

</details>

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `sigma = -0.8; w0 = 2*pi*2;`: parte real del exponente (amortiguamiento) y frecuencia angular.
- `x = exp((sigma + 1j*w0)*t);`: $x(t) = e^{(\sigma + j\omega_0)t}$. El resultado es un vector de **números complejos**.
- `plot(t, real(x), ...)` y `plot(t, imag(x), ...)`: partes real ($e^{\sigma t}\cos\omega_0 t$) e imaginaria ($e^{\sigma t}\sin\omega_0 t$).
- `plot(t, abs(x), '--', ...)` y `plot(t, -abs(x), ...)`: la magnitud $e^{\sigma t}$ es la **envolvente** que limita la oscilación.
- `plot3(t, real(x), imag(x), ...)`: gráfica en **3D**: `plot3(x, y, z)` recibe tres vectores del mismo tamaño. Aquí se ve la espiral que forma la señal compleja.
- `zlabel('Im')`: etiqueta del tercer eje.
- `view(35, 25)`: ángulo de la cámara: `view(azimut, elevación)` en grados.

</details>

<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

## 2.4 Biblioteca de señales con funciones anónimas

### Concepto
Si definimos las señales elementales como **funciones anónimas**, las operaciones del curso se escriben exactamente como en el papel: `x(t-2)`, `x(-t)`, `x(2*t)`.

Además, muchas señales por tramos se construyen **combinando** escalones y rampas:

$$x(t) = r(t) - r(t-1) - u(t-3) \;\Rightarrow\; \text{sube de 0 a 1 en } [0,1],\ \text{vale 1 en } [1,3),\ \text{luego } 0.$$

### Ejemplos

```matlab
% Ejemplo 2.4.1: nuestra "biblioteca" de señales elementales
u    = @(t) double(t >= 0);                 % escalón unitario
r    = @(t) t .* u(t);                      % rampa
rect = @(t) u(t + 0.5) - u(t - 0.5);        % pulso rectangular unitario
tri  = @(t) (1 - abs(t)) .* (abs(t) <= 1);  % pulso triangular
d    = @(n) double(n == 0);                 % impulso discreto

t = -2:0.001:2;
figure
subplot(2,2,1); plot(t, u(t), 'Color', navy);   grid on; title('u(t)');    axis([-2 2 -0.2 1.2])
subplot(2,2,2); plot(t, r(t), 'Color', navy);   grid on; title('r(t)')
subplot(2,2,3); plot(t, rect(t), 'Color', cian); grid on; title('\Pi(t)'); axis([-2 2 -0.2 1.2])
subplot(2,2,4); plot(t, tri(t), 'Color', cian);  grid on; title('\Lambda(t)'); axis([-2 2 -0.2 1.2])
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `u = @(t) double(t >= 0);`: escalón unitario como función: `u(t-2)`, `u(-t)`, etc. funcionan directamente.
- `r = @(t) t .* u(t);`: la rampa se define **usando** `u`. Una función anónima puede usar otras definidas antes.
- `rect = @(t) u(t + 0.5) - u(t - 0.5);`: pulso de ancho 1 como resta de dos escalones: uno que sube en −0.5 y otro que lo "apaga" en 0.5.
- `tri = @(t) (1 - abs(t)) .* (abs(t) <= 1);`: triángulo de altura 1 y base de −1 a 1. La máscara `(abs(t) <= 1)` lo anula fuera de ese intervalo.
- `d = @(n) double(n == 0);`: impulso discreto $\delta[n]$.
- `subplot(2,2,k)`: cuadrícula de 2×2 gráficas, numeradas 1-2 (arriba) y 3-4 (abajo).
- `title('\Lambda(t)')`: `\Lambda` produce la letra griega Λ.

</details>

```matlab
% Ejemplo 2.4.2: señal por tramos construida con rampas y escalones
x = @(t) r(t) - r(t-1) - u(t-3);
t = -1:0.001:5;
figure
plot(t, x(t), 'Color', navy); grid on; axis([-1 5 -0.3 1.3])
xlabel('t'); title('x(t) = r(t) - r(t-1) - u(t-3)')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `x = @(t) r(t) - r(t-1) - u(t-3);`: combina las funciones de la biblioteca.
  - `r(t)`: rampa que empieza a subir en $t = 0$.
  - `- r(t-1)`: una rampa que arranca en $t=1$ **cancela** la pendiente: desde ahí la señal queda constante en 1.
  - `- u(t-3)`: un escalón en $t=3$ resta 1: la señal vuelve a 0.
- `plot(t, x(t), ...)`: evalúa la función en todo el vector `t` y la grafica.

</details>

```matlab
% Ejemplo 2.4.3: tren de impulsos discreto como suma de impulsos desplazados
n = -2:12;
tren = zeros(size(n));
for k = 0:3:9
    tren = tren + d(n - k);      % suma de delta[n - k] para k = 0, 3, 6, 9
end
figure
stem(n, tren, 'filled', 'Color', cian); grid on; axis([-2 12 -0.2 1.2])
xlabel('n'); title('\Sigma \delta[n - 3k],  k = 0,...,3')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `n = -2:12;`: índices.
- `tren = zeros(size(n));`: empieza con una señal de ceros del mismo tamaño que `n`.
- `for k = 0:3:9`: `k` toma los valores 0, 3, 6 y 9.
- `tren = tren + d(n - k);`: `d(n-k)` es $\delta[n-k]$, un impulso desplazado a la posición `k`. Se va acumulando uno por vuelta.

</details>

## 2.5 Operaciones sobre la amplitud

### Concepto

| Operación | Expresión | Ejemplo físico |
|---|---|---|
| Escalamiento | $y(t) = A\,x(t)$ | Amplificador de ganancia $A$ |
| Desplazamiento de nivel (offset) | $y(t) = x(t) + B$ | Acondicionamiento de un sensor |
| Suma | $y(t) = x_1(t) + x_2(t)$ | Señal + ruido; superposición |
| Producto | $y(t) = x_1(t)\,x_2(t)$ | Modulación; "encender" una señal con $u(t)$ |

### Ejemplos

```matlab
% Ejemplo 2.5.1: suma -> señal útil + ruido de medición
rng(0);                                % semilla: el ruido es reproducible
t = 0:0.001:1;
limpia = sin(2*pi*3*t);
ruido  = 0.3*randn(size(t));           % ruido gaussiano de desviación 0.3
medida = limpia + ruido;
figure
plot(t, medida, 'Color', [0.7 0.7 0.7]); hold on
plot(t, limpia, 'Color', navy, 'LineWidth', 2); hold off; grid on
legend('medida = señal + ruido', 'señal útil'); xlabel('t [s]')
title('Suma de señales (filtrarla será el objetivo del Módulo 8)')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `rng(0);`: fija la **semilla** del generador de números aleatorios. Así todos obtienen el mismo "ruido" cada vez que ejecutan el bloque.
- `limpia = sin(2*pi*3*t);`: la señal útil, de 3 Hz.
- `ruido = 0.3*randn(size(t));`: `randn(size(t))` genera números aleatorios con distribución normal (media 0, desviación 1), uno por muestra. Al multiplicar por 0.3, la desviación queda en 0.3.
- `medida = limpia + ruido;`: la **suma** de señales es muestra a muestra.
- `plot(t, medida, 'Color', [0.7 0.7 0.7])`: el color también puede darse directamente como RGB; `[0.7 0.7 0.7]` es gris claro.

</details>

```matlab
% Ejemplo 2.5.2: producto -> modulación de amplitud (AM)
portadora = cos(2*pi*20*t);
envolvente = 1 + 0.5*cos(2*pi*2*t);
am = envolvente .* portadora;          % producto elemento a elemento
figure
plot(t, am, 'Color', navy); hold on
plot(t, envolvente, '--', 'Color', ambar); plot(t, -envolvente, '--', 'Color', ambar)
hold off; grid on; xlabel('t [s]'); title('Producto de señales: AM')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `portadora = cos(2*pi*20*t);`: senoidal rápida de 20 Hz.
- `envolvente = 1 + 0.5*cos(2*pi*2*t);`: variación lenta de 2 Hz, desplazada a 1 para que nunca sea negativa.
- `am = envolvente .* portadora;`: **producto** muestra a muestra: la amplitud de la portadora sigue a la envolvente (modulación AM).
- `plot(t, -envolvente, ...)`: se dibujan $\pm$ envolvente para ver que la señal queda "encerrada".

</details>

```matlab
% Ejemplo 2.5.3: escalamiento y offset -> acondicionamiento de un sensor
% Sensor de temperatura: 0-5 V corresponden a 0-100 °C  => T = 20*V
V = 2.5 + 1.5*sin(2*pi*0.5*t);       % voltaje del sensor [V]
T = 20*V;                            % escalamiento: ganancia 20 °C/V
Vadc = 0.66*V - 0.5;                 % ganancia + offset para otro rango de ADC
figure
subplot(2,1,1); plot(t, V, 'Color', navy); grid on; ylabel('V [V]'); title('Salida del sensor')
subplot(2,1,2); plot(t, T, 'Color', cian); grid on; ylabel('T [°C]'); xlabel('t [s]')
title('Señal escalada: T = 20 V')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `V = 2.5 + 1.5*sin(2*pi*0.5*t);`: voltaje del sensor, una oscilación de 0.5 Hz entre 1 y 4 V.
- `T = 20*V;`: **escalamiento** de amplitud: multiplica por la ganancia 20 °C/V para obtener temperatura.
- `Vadc = 0.66*V - 0.5;`: ganancia (0.66) más **offset** (−0.5): cambia la escala y desplaza el nivel para adaptarlo a otro rango de entrada.
- `ylabel('T [°C]')`: etiqueta con unidades.

</details>

```matlab
% Ejemplo 2.5.4: producto con un escalón -> "encender" una señal en t = 0.3 s
t = -0.2:0.001:1;
y = sin(2*pi*5*t) .* u(t - 0.3);
figure
plot(t, y, 'Color', navy); grid on; xlabel('t [s]')
title('sin(10\pi t) u(t - 0.3): la señal aparece en t = 0.3 s')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

</details>

## 2.6 Operaciones sobre la variable independiente (tiempo continuo)

### Concepto

| Operación | Expresión | Efecto |
|---|---|---|
| Desplazamiento | $x(t - t_0)$ | $t_0>0$: **retraso** (a la derecha); $t_0<0$: adelanto (a la izquierda) |
| Inversión | $x(-t)$ | Reflejo respecto al eje vertical |
| Escalamiento | $x(at)$ | $\lvert a\rvert>1$: **compresión**; $\lvert a\rvert<1$: expansión |
| Combinada | $x(at - b)$ | Primero desplazar por $b$ y después escalar por $a$ |

**Orden seguro para $y(t) = x(at-b)$:**
1. Desplazar: $v(t) = x(t - b)$.
2. Escalar o invertir: $y(t) = v(at) = x(at - b)$ ✔.

> 💡 **Analogía:** grabas un video (la señal). Desplazar es **empezar a reproducirlo más tarde**, invertir es **verlo al revés** y escalar es **cambiar la velocidad de reproducción** (×2 dura la mitad).

Con funciones anónimas MATLAB lo hace directamente: `plot(t, x(a*t - b))`. Usamos una señal **asimétrica** para que el efecto se vea con claridad:

$$x(t) = r(t) - r(t-1) - u(t-2) = \begin{cases} t & 0\le t<1\\ 1 & 1\le t<2 \\ 0 & \text{otro caso}\end{cases}$$

### Ejemplos

```matlab
% Señal base asimétrica para todos los ejemplos de este ítem
x = @(t) r(t) - r(t-1) - u(t-2);
t = -5:0.001:5;
figure
plot(t, x(t), 'Color', navy); grid on; axis([-5 5 -0.3 1.3])
xlabel('t'); title('Señal base x(t)')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `x = @(t) r(t) - r(t-1) - u(t-2);`: la señal base: sube de 0 a 1 entre $t=0$ y $t=1$, vale 1 hasta $t=2$ y luego vuelve a 0. Es **asimétrica** a propósito, para que se note si quedó invertida o desplazada.
- `t = -5:0.001:5;`: eje de tiempo amplio para que quepan los desplazamientos.
- `axis([-5 5 -0.3 1.3])`: mismos límites en todas las gráficas de este ítem, para poder compararlas.

</details>

```matlab
% Ejemplo 2.6.1: desplazamientos x(t-2) (retraso) y x(t+1) (adelanto)
figure
subplot(3,1,1); plot(t, x(t), 'Color', navy);  grid on; axis([-5 5 -0.3 1.3]); title('x(t)')
subplot(3,1,2); plot(t, x(t-2), 'Color', cian); grid on; axis([-5 5 -0.3 1.3]); title('x(t - 2): se mueve 2 a la DERECHA')
subplot(3,1,3); plot(t, x(t+1), 'Color', ambar); grid on; axis([-5 5 -0.3 1.3]); title('x(t + 1): se mueve 1 a la IZQUIERDA'); xlabel('t')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `plot(t, x(t-2), ...)`: al pasar `t-2` como argumento, se grafica $x(t-2)$. Cada valor de la señal aparece **2 unidades más tarde**: desplazamiento a la derecha (retraso).

</details>

- `x(-t+2)`: se puede leer como $x(-(t-2))$: la señal reflejada y luego **desplazada 2 a la derecha**. Ocupa el intervalo $[0, 2]$.

</details>

</details>

## 2.7 Operaciones en tiempo discreto

### Concepto
Las operaciones son análogas a las de CT, pero $n$ **debe ser entero**:

| Operación | Expresión | Particularidad |
|---|---|---|
| Desplazamiento | $x[n - n_0]$ | $n_0$ entero |
| Inversión | $x[-n]$ | Igual que en CT |
| **Diezmado** | $x[Mn]$ | Conserva una de cada $M$ muestras y **se pierde información** |
| **Interpolación** (expansión) | $x[n/L]$ | Solo existe cuando $n/L$ es entero; en los demás puntos vale **0** |

Hay dos formas de trabajar en MATLAB:
- **Con funciones anónimas**, igual que en CT: `x(n-3)`, `x(-n)`.
- **Con datos (vectores)**, cuando la señal viene de una medición: se transforma el **eje** `n` y se conservan los valores.

### Ejemplos

```matlab
% Ejemplo 2.7.1: desplazamiento e inversión con función anónima
xd = @(n) (n + 1) .* (n >= 0 & n <= 4);     % x[n] = {1,2,3,4,5} para n = 0..4
n = -8:8;
figure
subplot(3,1,1); stem(n, xd(n), 'filled', 'Color', navy);   grid on; title('x[n]')
subplot(3,1,2); stem(n, xd(n-3), 'filled', 'Color', cian); grid on; title('x[n - 3]: retraso de 3 muestras')
subplot(3,1,3); stem(n, xd(-n), 'filled', 'Color', ambar); grid on; title('x[-n]: inversión'); xlabel('n')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `xd = @(n) (n + 1) .* (n >= 0 & n <= 4);`: señal discreta que vale `n+1` (1, 2, 3, 4, 5) para $0 \le n \le 4$ y 0 en otro caso. La máscara `(n >= 0 & n <= 4)` hace el "recorte".
- `n = -8:8;`: índices enteros.
- `stem(n, xd(n-3), ...)`: $x[n-3]$, retraso de 3 muestras.
- `stem(n, xd(-n), ...)`: $x[-n]$, inversión.

</details>

```matlab
% Ejemplo 2.7.2: diezmado x[2n] -> se pierden muestras
figure
subplot(2,1,1); stem(n, xd(n), 'filled', 'Color', navy); grid on; title('x[n]: 5 muestras no nulas')
subplot(2,1,2); stem(n, xd(2*n), 'filled', 'Color', cian); grid on
title('x[2n]: solo sobreviven x[0], x[2], x[4]'); xlabel('n')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `stem(n, xd(2*n), ...)`: $x[2n]$, **diezmado**. Para cada `n` se toma la muestra de la posición `2n`: $x[0], x[2], x[4], \dots$. Las muestras impares $x[1]$ y $x[3]$ **se pierden** y no se pueden recuperar.

</details>

```matlab
% Ejemplo 2.7.3: interpolación x[n/2] -> se insertan ceros
es_entero = (mod(n, 2) == 0);                % n/2 es entero solo si n es par
y_interp = xd(n/2) .* es_entero;             % fuera de los pares la señal vale 0
figure
stem(n, y_interp, 'filled', 'Color', ambar); grid on; xlabel('n')
title('x[n/2]: muestras originales separadas por ceros')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `es_entero = (mod(n, 2) == 0);`: `mod(n, 2)` es el residuo al dividir entre 2; vale 0 si `n` es par. Así `es_entero` marca con 1 los `n` donde $n/2$ es entero.
- `y_interp = xd(n/2) .* es_entero;`: evalúa $x[n/2]$ y lo multiplica por la máscara, de modo que en los `n` impares (donde $n/2$ no es entero y la señal no está definida) queda 0. Resultado: las muestras originales separadas por ceros.

</details>

```matlab
% Ejemplo 2.7.4: trabajar con DATOS (vector de valores + vector de índices)
nx = 0:4;            % índices donde está definida la señal
vx = [1 2 3 2 1];    % valores medidos
figure
subplot(3,1,1); stem(nx, vx, 'filled', 'Color', navy); grid on; axis([-6 8 0 3.5]); title('x[n]')
% Desplazamiento x[n-3]: los valores no cambian, el eje se mueve +3
subplot(3,1,2); stem(nx + 3, vx, 'filled', 'Color', cian); grid on; axis([-6 8 0 3.5]); title('x[n - 3]')
% Inversión x[-n]: se reflejan el eje y el orden de los valores
subplot(3,1,3); stem(-fliplr(nx), fliplr(vx), 'filled', 'Color', ambar); grid on; axis([-6 8 0 3.5])
title('x[-n]'); xlabel('n')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `nx = 0:4;` y `vx = [1 2 3 2 1];`: la señal viene como **datos**: un vector de posiciones y otro de valores.
- `stem(nx + 3, vx, ...)`: para retrasar 3 muestras basta **sumar 3 al eje**; los valores no cambian.
- `stem(-fliplr(nx), fliplr(vx), ...)`: para invertir:
  - `fliplr` (*flip left-right*) invierte el orden de un vector fila: `[0 1 2 3 4]` → `[4 3 2 1 0]`.
  - `-fliplr(nx)` da el nuevo eje `[-4 ... 0]`, y `fliplr(vx)` reordena los valores para que coincidan.

</details>

## 2.8 Periodicidad

### Concepto
- **CT:** $x(t)$ es periódica si existe $T>0$ tal que $x(t + T) = x(t)$ para todo $t$. El menor $T$ es el **periodo fundamental** $T_0$. La suma de dos senoidales es periódica solo si $T_1/T_2$ es **racional**; entonces $T_0 = \mathrm{mcm}(T_1, T_2)$.
- **DT:** $x[n]$ es periódica si existe un **entero** $N>0$ tal que $x[n+N] = x[n]$. Para $\cos(\Omega_0 n)$:

$$\text{periódica} \iff \frac{\Omega_0}{2\pi} = \frac{k}{N}\ \text{es racional}, \qquad N = \text{el menor entero con } \Omega_0 N = 2\pi k.$$

⚠️ ¡Sorpresa! $\cos(n)$ **no** es periódica, porque $\frac{1}{2\pi}$ es irracional, aunque $\cos(t)$ sí lo sea.

### Ejemplos

```matlab
% Ejemplo 2.8.1: suma de senoidales con periodos conmensurables
% x(t) = cos(2*pi*2t) + sin(2*pi*3t): T1 = 1/2 s, T2 = 1/3 s  ->  T0 = 1 s
xp = @(t) cos(2*pi*2*t) + sin(2*pi*3*t);
t = 0:0.0005:3;
T0 = 1;
error_T0 = max(abs(xp(t + T0) - xp(t)))        % ~0  => periódica con T0 = 1
error_05 = max(abs(xp(t + 0.5) - xp(t)))       % ≠ 0 => 0.5 s NO es periodo
figure
plot(t, xp(t), 'Color', navy); grid on; xlabel('t [s]')
title('cos(4\pi t) + sin(6\pi t): se repite cada T_0 = 1 s')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `xp = @(t) cos(2*pi*2*t) + sin(2*pi*3*t);`: suma de senoidales de 2 Hz y 3 Hz.
- `T0 = 1;`: periodo propuesto: mcm de $1/2$ y $1/3$ = 1 s.
- `error_T0 = max(abs(xp(t + T0) - xp(t)))`: compara la señal con ella misma desplazada `T0`. Si la mayor diferencia es prácticamente 0 (≈ $10^{-14}$, error de redondeo), entonces $x(t+T_0) = x(t)$: es periódica con ese periodo.
- `error_05 = ...`: la misma prueba con 0.5 s da una diferencia grande (2): 0.5 s **no** es periodo.

</details>

```matlab
% Ejemplo 2.8.2: suma NO periódica: cos(2t) + cos(pi t)  (T1/T2 = 2/pi, irracional)
xnp = @(t) cos(2*t) + cos(pi*t);
t = 0:0.005:60;
figure
plot(t, xnp(t), 'Color', ambar); grid on; xlabel('t [s]')
title('cos(2t) + cos(\pi t): nunca se repite exactamente')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `xnp = @(t) cos(2*t) + cos(pi*t);`: frecuencias angulares 2 y π rad/s. Su cociente $2/\pi$ es irracional, por lo que no existe un periodo común.
- `t = 0:0.005:60;`: 60 s de señal para comprobar visualmente que el patrón nunca se repite exactamente.

</details>

```matlab
% Ejemplo 2.8.3: periodicidad en DT: cos(pi n/4) sí, cos(n) no
n = 0:40;
figure
subplot(2,1,1); stem(n, cos(pi*n/4), 'filled', 'Color', navy); grid on
title('cos(\pi n/4): periódica, N = 8')
subplot(2,1,2); stem(n, cos(n), 'filled', 'Color', ambar); grid on
title('cos(n): NO periódica (\Omega_0/2\pi = 1/2\pi es irracional)'); xlabel('n')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `stem(n, cos(pi*n/4), ...)`: $\Omega_0 = \pi/4$, y $\Omega_0/2\pi = 1/8$ es racional: periódica con $N = 8$.
- `stem(n, cos(n), ...)`: $\Omega_0 = 1$, y $1/2\pi$ es irracional: **no** es periódica, aunque las muestras sigan una forma "ondulada".

</details>

```matlab
% Ejemplo 2.8.4: buscar numéricamente el periodo fundamental de una señal DT
xdt = @(n) cos(3*pi*n/5);           % Omega0/2pi = 3/10  =>  N = 10
n = 0:200;
N0 = NaN;
for N = 1:100
    if max(abs(xdt(n + N) - xdt(n))) < 1e-9
        N0 = N;                      % primer N que cumple: el periodo fundamental
        break
    end
end
N0
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `xdt = @(n) cos(3*pi*n/5);`: señal a analizar.
- `N0 = NaN;`: valor inicial "sin resultado", por si no se encuentra periodo.
- `for N = 1:100`: prueba cada candidato `N` de 1 a 100.
- `if max(abs(xdt(n + N) - xdt(n))) < 1e-9`: si la señal desplazada `N` muestras es igual a la original (diferencia menor que $10^{-9}$, tolerancia al redondeo), `N` es un periodo.
- `N0 = N; break`: guarda el **primer** `N` que cumple (el periodo fundamental) y `break` sale del bucle de inmediato.

</details>

## 2.9 Componentes par e impar, energía y potencia

### Concepto
**Simetría:** $x(t)$ es **par** si $x(-t) = x(t)$ e **impar** si $x(-t) = -x(t)$. Toda señal se descompone como

$$x(t) = x_e(t) + x_o(t), \qquad x_e(t) = \frac{x(t) + x(-t)}{2}, \qquad x_o(t) = \frac{x(t) - x(-t)}{2}.$$

**Energía y potencia:**

| | Tiempo continuo | Tiempo discreto |
|---|---|---|
| Energía | $E = \displaystyle\int_{-\infty}^{\infty} \lvert x(t)\rvert^2 dt$ | $E = \displaystyle\sum_{n=-\infty}^{\infty} \lvert x[n]\rvert^2$ |
| Potencia | $P = \displaystyle\lim_{T\to\infty}\frac{1}{2T}\int_{-T}^{T} \lvert x(t)\rvert^2 dt$ | $P = \displaystyle\lim_{N\to\infty}\frac{1}{2N+1}\sum_{-N}^{N}\lvert x[n]\rvert^2$ |

- **Señal de energía:** $0 < E < \infty$ (y $P = 0$). Ejemplo: pulsos y exponenciales decrecientes.
- **Señal de potencia:** $0 < P < \infty$ (y $E = \infty$). Ejemplo: senoidales y el escalón. Para una señal periódica basta promediar **un periodo**: $P = \frac{1}{T_0}\int_{T_0}\lvert x\rvert^2dt$.

En MATLAB la integral se aproxima con `trapz(t, y)` y la suma con `sum`.

### Ejemplos

```matlab
% Ejemplo 2.9.1: componentes par e impar de x(t) = e^{-t} u(t)
xa = @(t) exp(-t) .* u(t);
xe = @(t) (xa(t) + xa(-t))/2;       % componente par  = e^{-|t|}/2
xo = @(t) (xa(t) - xa(-t))/2;       % componente impar
t = -4:0.001:4;
figure
subplot(3,1,1); plot(t, xa(t), 'Color', navy); grid on; title('x(t) = e^{-t}u(t)')
subplot(3,1,2); plot(t, xe(t), 'Color', cian); grid on; title('x_e(t): par')
subplot(3,1,3); plot(t, xo(t), 'Color', ambar); grid on; title('x_o(t): impar'); xlabel('t')
reconstruccion = max(abs(xe(t) + xo(t) - xa(t)))   % ~0: x = xe + xo
% Nota: en t = 0, x_e muestra un punto aislado en 1 porque u(0) = 1 se suma dos veces
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `xa = @(t) exp(-t) .* u(t);`: $x(t) = e^{-t}u(t)$.
- `xe = @(t) (xa(t) + xa(-t))/2;`: componente **par**, directamente de la fórmula $\frac{x(t)+x(-t)}{2}$.
- `xo = @(t) (xa(t) - xa(-t))/2;`: componente **impar**, $\frac{x(t)-x(-t)}{2}$.
- `reconstruccion = max(abs(xe(t) + xo(t) - xa(t)))`: verifica que $x_e + x_o = x$; da 0.

</details>

```matlab
% Ejemplo 2.9.2: componentes par e impar en DT
xn = @(n) (n + 1) .* (n >= 0 & n <= 3);      % {1, 2, 3, 4} en n = 0..3
n = -5:5;
xen = (xn(n) + xn(-n))/2;
xon = (xn(n) - xn(-n))/2;
figure
subplot(3,1,1); stem(n, xn(n), 'filled', 'Color', navy); grid on; title('x[n]')
subplot(3,1,2); stem(n, xen, 'filled', 'Color', cian);   grid on; title('x_e[n]')
subplot(3,1,3); stem(n, xon, 'filled', 'Color', ambar);  grid on; title('x_o[n]'); xlabel('n')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `xn = @(n) (n + 1) .* (n >= 0 & n <= 3);`: la señal {1, 2, 3, 4} en $n = 0, \dots, 3$.
- `xen = (xn(n) + xn(-n))/2;` y `xon = (xn(n) - xn(-n))/2;`: las mismas fórmulas de CT, ahora con índices enteros. Aquí se guardan como **vectores** (ya evaluados) en lugar de funciones.

</details>

```matlab
% Ejemplo 2.9.3: energía de x(t) = e^{-2t} u(t)  -> analítica: E = 1/4
t = 0:0.0001:10;                         % 10 s es "infinito" para e^{-2t}
E_num = trapz(t, abs(exp(-2*t)).^2)
E_teo = 1/4
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `t = 0:0.0001:10;`: el límite superior de la integral es ∞, pero $e^{-2t}$ es prácticamente 0 después de pocos segundos; 10 s es suficiente. El paso pequeño mejora la precisión.
- `E_num = trapz(t, abs(exp(-2*t)).^2)`: calcula $\int |x(t)|^2 dt$ de forma numérica.
  - `abs(...)`: magnitud (necesaria si la señal fuera compleja).
  - `.^2`: al cuadrado, elemento a elemento.
  - `trapz(t, y)`: área bajo la curva `y` sobre el eje `t`.
- `E_teo = 1/4`: valor analítico, para comparar.

</details>

```matlab
% Ejemplo 2.9.4: potencia de x(t) = 3 cos(2*pi*5 t)  -> analítica: P = A^2/2 = 4.5
T0 = 1/5;
t = 0:T0/1000:T0;                        % basta un periodo
P_num = (1/T0) * trapz(t, (3*cos(2*pi*5*t)).^2)
P_teo = 3^2/2
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `T0 = 1/5;`: periodo de una senoidal de 5 Hz.
- `t = 0:T0/1000:T0;`: un solo periodo, dividido en 1000 pasos.
- `P_num = (1/T0) * trapz(t, (3*cos(2*pi*5*t)).^2)`: potencia promedio $\frac{1}{T_0}\int_{T_0} |x|^2 dt$: el área de $x^2$ en un periodo dividida entre la duración del periodo.
- `P_teo = 3^2/2`: para una senoidal de amplitud $A$, $P = A^2/2 = 4.5$.

</details>

```matlab
% Ejemplo 2.9.5: energía en DT: x[n] = 0.5^n u[n] -> E = 1/(1 - 0.25) = 4/3
n = 0:100;
[E_dt, ~] = energia_potencia(0.5.^n)     % reutilizamos la función del ítem 1.7
E_teo = 4/3
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `n = 0:100;`: la suma va hasta ∞, pero $0.5^n$ es despreciable mucho antes de $n = 100$.
- `[E_dt, ~] = energia_potencia(0.5.^n)`: reutiliza la función creada en 1.7.
  - `0.5.^n`: la señal $x[n] = 0.5^n$.
  - `~` (tilde) en la lista de salidas significa "esta salida no me interesa": se **descarta** la potencia y solo se guarda la energía.
- `E_teo = 4/3`: serie geométrica $\sum (0.25)^n = \frac{1}{1-0.25}$.

</details>

## 2.10 Aplicación mecatrónica: PWM y encoder

### Concepto
**PWM (modulación por ancho de pulso).** Es una señal cuadrada de periodo $T$ y **ciclo de trabajo** $D$:

$$v_{PWM}(t) = \begin{cases} V_{cc} & 0 \le \bmod(t, T) < DT \\ 0 & \text{otro caso}\end{cases} \qquad\Rightarrow\qquad \bar v = D\,V_{cc}.$$

> 💡 **Analogía:** es como abrir y cerrar un **grifo** muy rápido para llenar un tanque. El tanque, que aquí es la inercia del motor, no "ve" cada apertura: ve el **caudal promedio**. Por eso el motor gira como si recibiera $D\cdot V_{cc}$.

**Encoder incremental en cuadratura.** Entrega dos señales cuadradas, **A** y **B**, desfasadas 90°. Con $N$ pulsos por revolución (PPR):
- Velocidad: $\text{rpm} = \dfrac{\text{pulsos contados}}{N \cdot T_{ventana}}\cdot 60$.
- Sentido de giro: en cada flanco de subida de A, el valor de B indica la dirección.

### Ejemplos

```matlab
% Ejemplo 2.10.1: señales PWM con distintos ciclos de trabajo
Vcc = 12;  fpwm = 1000;  T = 1/fpwm;       % 12 V, 1 kHz
t = 0:1e-6:3*T;                             % 3 periodos con resolución de 1 us
pwm = @(D) Vcc * double(mod(t, T) < D*T);   % generador de PWM (función de D)
D = [0.25 0.50 0.75];
figure
for k = 1:3
    subplot(3,1,k)
    plot(t*1e3, pwm(D(k)), 'Color', navy); hold on
    plot(t*1e3, mean(pwm(D(k)))*ones(size(t)), '--', 'Color', ambar); hold off
    grid on; axis([0 3 -1 13]); ylabel('V')
    title(sprintf('D = %d %%  ->  promedio = %.2f V', D(k)*100, mean(pwm(D(k)))))
end
xlabel('t [ms]')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `Vcc = 12; fpwm = 1000; T = 1/fpwm;`: tensión de alimentación, frecuencia del PWM (1 kHz) y su periodo (1 ms).
- `t = 0:1e-6:3*T;`: tres periodos con resolución de 1 µs (`1e-6`), para que los flancos se vean nítidos.
- `pwm = @(D) Vcc * double(mod(t, T) < D*T);`: función anónima cuya entrada es el **ciclo de trabajo** `D`.
  - `mod(t, T)`: tiempo dentro del periodo actual (reinicia en 0 cada `T`).
  - `< D*T`: vale 1 durante la fracción `D` del periodo.
  - `Vcc * ...`: la convierte en voltios.
  - Nota: esta función usa el vector `t` que existía al momento de definirla.
- `D = [0.25 0.50 0.75];`: tres ciclos de trabajo.
- `plot(t*1e3, pwm(D(k)), ...)`: `t*1e3` pasa el eje a **milisegundos** para leerlo mejor.
- `mean(pwm(D(k)))*ones(size(t))`: el promedio de la señal (un número) se repite con `ones` para dibujarlo como línea horizontal.
- `sprintf('D = %d %%  ->  promedio = %.2f V', ...)`: `%d` imprime un entero, `%%` imprime el símbolo %, y `%.2f` un número con 2 decimales.

</details>

```matlab
% Ejemplo 2.10.2: la inercia del motor "promedia" el PWM (sistema de primer orden)
% Modelo discreto (Euler): w[n] = w[n-1] + (Ts/tau) * (v[n] - w[n-1])
Ts = 1e-6;  tau = 5e-3;                      % constante de tiempo del motor: 5 ms
t = 0:Ts:40e-3;                              % 40 ms de simulación
v = Vcc * double(mod(t, T) < 0.5*T);         % PWM al 50 %
w = zeros(size(t));                          % "velocidad" en voltios equivalentes
for k = 2:numel(t)
    w(k) = w(k-1) + (Ts/tau)*(v(k) - w(k-1));
end
figure
plot(t*1e3, v, 'Color', [0.8 0.8 0.8]); hold on
plot(t*1e3, w, 'Color', navy, 'LineWidth', 2); hold off; grid on
xlabel('t [ms]'); ylabel('V'); legend('PWM (entrada)', 'respuesta del motor')
title('El motor responde al valor promedio: D V_{cc} = 6 V')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `Ts = 1e-6; tau = 5e-3;`: paso de simulación (1 µs) y constante de tiempo mecánica del motor (5 ms).
- `t = 0:Ts:40e-3;`: 40 ms de simulación.
- `v = Vcc * double(mod(t, T) < 0.5*T);`: PWM al 50 % (la misma idea del ejemplo anterior).
- `w = zeros(size(t));`: reserva memoria para la respuesta.
- `for k = 2:numel(t)`: empieza en 2 porque cada valor usa el anterior, `w(k-1)`.
- `w(k) = w(k-1) + (Ts/tau)*(v(k) - w(k-1));`: modelo de primer orden discretizado (método de Euler). En cada paso, la salida se acerca a la entrada una fracción `Ts/tau`. Un sistema lento **no alcanza** a seguir los pulsos y se queda en el promedio.

</details>

```matlab
% Ejemplo 2.10.3: señales A y B de un encoder en cuadratura
N   = 20;                 % pulsos por revolución (PPR)
rpm = 120;                % velocidad del eje
fs  = 1e5;  t = 0:1/fs:0.1;
theta = 2*pi*(rpm/60)*t;                  % posición angular [rad]
p = N*theta/(2*pi);                       % número de pulsos recorridos
A = double(mod(p, 1) < 0.5);              % canal A
B = double(mod(p - 0.25, 1) < 0.5);       % canal B: retrasado 1/4 de ciclo (90°)
figure
subplot(2,1,1); plot(t*1e3, A, 'Color', navy); grid on; axis([0 100 -0.2 1.2]); title('Canal A')
subplot(2,1,2); plot(t*1e3, B, 'Color', cian); grid on; axis([0 100 -0.2 1.2]); title('Canal B (desfase 90°)')
xlabel('t [ms]')
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `N = 20;`: resolución del encoder: 20 pulsos por revolución (PPR).
- `rpm = 120;`: velocidad del eje (2 vueltas por segundo).
- `fs = 1e5; t = 0:1/fs:0.1;`: muestreo a 100 kHz durante 0.1 s.
- `theta = 2*pi*(rpm/60)*t;`: posición angular en radianes; `rpm/60` convierte a vueltas por segundo.
- `p = N*theta/(2*pi);`: número de pulsos recorridos (vueltas × PPR).
- `A = double(mod(p, 1) < 0.5);`: `mod(p, 1)` es la parte fraccionaria de `p`; el canal A vale 1 durante la primera mitad de cada pulso (señal cuadrada al 50 %).
- `B = double(mod(p - 0.25, 1) < 0.5);`: igual que A pero desplazado **un cuarto de pulso** (90°): eso es lo que significa "cuadratura".

</details>

```matlab
% Ejemplo 2.10.4: estimar velocidad y sentido de giro a partir de A y B
Tv = 1;  t = 0:1/fs:Tv;                    % ventana de medición de 1 s
for sentido = [1 -1]                       % +1 horario, -1 antihorario
    p = sentido * N*(rpm/60)*t;
    A = double(mod(p, 1) < 0.5);
    B = double(mod(p - 0.25, 1) < 0.5);
    flancos = find(diff(A) == 1);          % flancos de subida de A
    rpm_est = numel(flancos)/(N*Tv)*60;    % pulsos / (PPR * ventana) * 60
    if mean(B(flancos + 1)) < 0.5          % B = 0 en el flanco => A adelanta a B
        sentido_txt = 'horario';
    else
        sentido_txt = 'antihorario';
    end
    fprintf('Velocidad estimada: %.0f rpm, sentido %s\n', rpm_est, sentido_txt)
end
```

<details>
<summary><b>🔍 ¿Qué hace cada línea?</b> <i>(clic para desplegar)</i></summary>

- `Tv = 1; t = 0:1/fs:Tv;`: ventana de medición de 1 s.
- `for sentido = [1 -1]`: repite para giro horario (+1) y antihorario (−1).
- `p = sentido * N*(rpm/60)*t;`: con signo negativo, la posición disminuye (gira al revés).
- `flancos = find(diff(A) == 1);`: `diff(A)` vale +1 justo donde A pasa de 0 a 1 (**flanco de subida**); `find` devuelve esas posiciones.
- `rpm_est = numel(flancos)/(N*Tv)*60;`: pulsos contados ÷ (PPR × tiempo) = vueltas por segundo; ×60 = rpm.
- `if mean(B(flancos + 1)) < 0.5`: mira cuánto vale B en cada flanco de subida de A (`flancos + 1` es la primera muestra después del flanco) y promedia. Si B suele estar en 0, A va **adelante** de B: sentido horario.
- `fprintf('... %.0f rpm, sentido %s\n', rpm_est, sentido_txt)`: `%.0f` muestra el número sin decimales y `%s` inserta un texto.

</details>
